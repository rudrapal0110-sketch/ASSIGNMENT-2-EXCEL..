Q8. Select independent variables (features) and dependent variable (target).
 Target variable: medv (house price)

```python
#INDEPENDENT VARIABLES
X = df.drop('medv', axis=1)
```

```python
#DEPENDENT VARIABLE
y = df['medv']
```

Q9. Split the dataset into training and testing sets (80% training, 20% testing).


```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

Q10. Train a Linear Regression model and perform the following:

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X_train, y_train)
# MAKE PREDICTIONS
predictions = model.predict(X_test)
```

```python
# CALCULATE MAE
from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(
    y_test,
    predictions
)

print(mae)
```

```python
#CALCULATE MSE
from sklearn.metrics import mean_squared_error

mse = mean_squared_error(
    y_test,
    predictions
)

print(mse)
```

```python
# CALCULATE RMSE
import numpy as np

rmse = np.sqrt(mse)

print(rmse)
```

```python
# CALCULATE R² SCORE
from sklearn.metrics import r2_score

r2 = r2_score(
    y_test,
    predictions
)

print(r2)
```

```python
# CALCULATE ADJUSTED R²
n = X_test.shape[0]

k = X_test.shape[1]

adjusted_r2 = 1 - (
    (1-r2)*(n-1)
) / (
    n-k-1
)

print(adjusted_r2)
```
